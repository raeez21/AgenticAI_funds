```python
from smolagents import ToolCallingAgent, WebSearchTool, LiteLLMModel
```

```python
model  = LiteLLMModel(model_id = 'ollama_chat/qwen-tools',
                      api_base='http://localhost:11434')

# here qwen-tools is a new model we build from qwen2
```

```python
agent = ToolCallingAgent(
    tools=[WebSearchTool()],
    model=model
)

agent.run("Search for the best music recommendations for a party at Wayne's mansion.")
```
